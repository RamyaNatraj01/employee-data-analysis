## Task 2: Exploratory Data Analysis (EDA)

**Objective:**

To perform exploratory data analysis on the cleaned employee dataset
using summary statistics, groupby, correlation, business questions,
data merging and pivot tables.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("clean_data.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

## 1. Summary Statistics

Calculate the following statistical measures for numerical columns:

- Mean
- Median
- Minimum
- Maximum
- Standard Deviation

These statistics help us understand the distribution
and characteristics of employee data.

```python
numeric_columns = df.select_dtypes(include=np.number).columns

summary_stats = df[numeric_columns].agg(
    ['mean', 'median', 'min', 'max', 'std']
)

summary_stats
```

## 2. GroupBy Analysis

Use groupby().agg() to compare employee information
across different departments.

We calculate:
- Average salary
- Maximum salary
- Minimum salary
- Average age
- Employee count

```python
department_analysis = df.groupby('Department').agg(
    Average_Age=('Age', 'mean'),
    Maximum_Age=('Age', 'max'),
    Minimum_Age=('Age', 'min'),
    Average_DailyRate=('DailyRate', 'mean'),
    Average_DistanceFromHome=('DistanceFromHome', 'mean'),
    Employee_Count=('Age', 'count')
)

department_analysis
```

```python
department_analysis = df.groupby('Department').agg(
    Average_Age=('Age', 'mean'),
    Average_DailyRate=('DailyRate', 'mean'),
    Average_DistanceFromHome=('DistanceFromHome', 'mean'),
    Employee_Count=('Age', 'count')
)

department_analysis
```

```python
jobrole_analysis = df.groupby('JobRole').agg(
    Average_Age=('Age', 'mean'),
    Average_DailyRate=('DailyRate', 'mean'),
    Average_MonthlyIncome=('MonthlyIncome', 'mean'),
    Employee_Count=('Age', 'count')
)

jobrole_analysis
```

## 3. Correlation Analysis

Correlation helps us understand the relationship
between numerical variables.

We use df.corr() to calculate correlation coefficients.

Correlation values range from -1 to +1.

- +1 indicates a perfect positive relationship.
- -1 indicates a perfect negative relationship.
- 0 indicates no linear relationship.

```python
correlation_matrix = df.select_dtypes(
    include=np.number
).corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
corr_pairs = correlation_matrix.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]

corr_pairs = corr_pairs.drop_duplicates()

top_3_correlations = corr_pairs.abs().sort_values(
    ascending=False
).head(3)

top_3_correlations
```

## 4. Answering Business Questions

We will answer five business questions using
numerical analysis.

1. Which department has the highest average salary?
2. Which city has the highest number of employees?
3. What is the average salary of employees?
4. Which gender has the highest average salary?
5. What is the average age of employees?

These questions help management understand
employee distribution and salary patterns.

```python
# Which department has the highest average salary?
```

```python
# Which city has the highest number of employees?

city_counts = df['City'].value_counts()

highest_city = city_counts.idxmax()
employee_count = city_counts.max()

print("City with highest employees:", highest_city)
print("Number of employees:", employee_count)
```

```python
# What is the average salary of employees?

average_salary = df['Salary'].mean()

print("Average employee salary:", round(average_salary, 2))
```

```python
# Which gender has the highest average salary?

gender_salary = df.groupby('Gender')['Salary'].mean()

highest_gender = gender_salary.idxmax()
highest_gender_salary = gender_salary.max()

print("Gender with highest average salary:", highest_gender)
print("Average salary:", round(highest_gender_salary, 2))
```

```python
# What is the average age of employees?

average_age = df['Age'].mean()

print("Average employee age:", round(average_age, 2))
```

## 5. Merging Department Budget Data

Create a second small dataset containing department budgets.

Use pd.merge() to combine the employee dataset
with the department budget dataset.

This helps us compare employee information
with department-level budget allocation.

```python
department_budget = pd.DataFrame({
    'Department': [
        'Hr',
        'Finance',
        'It',
        'Sales',
        'Marketing'
    ],
    'Budget': [
        500000,
        750000,
        1000000,
        850000,
        600000
    ]
})

department_budget
```

```python
merged_df = pd.merge(
    df,
    department_budget,
    on='Department',
    how='left'
)

merged_df.head()
```

## 6. Bonus Challenge: Pivot Table

Create a pivot table using two categorical columns.

We use Department and Gender to compare
employee counts across different groups.

```python
pivot_table = pd.pivot_table(
    df,
    index='Department',
    columns='Gender',
    values='Age',
    aggfunc='count',
    fill_value=0,
    observed=False
)

pivot_table
```
